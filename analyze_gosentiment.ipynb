```python
from json import load, dump
```

```python
lst = []
with open("go-sentiment.json", "r", encoding="utf-8") as file :
    lst = list(load(file))
```

```python
len(lst)
```

```python
lst[0]
```

```python
max_sent = {
    "pos" : 0,
    "neg" : 0,
    "neu" : 0,
    "amb" : 0
}
```

```python
# finding the max values in the sentiments
for data in lst :
    max_sent["pos"] = max(max_sent["pos"], data["senti"]["pos"])
    max_sent["neg"] = max(max_sent["neg"], data["senti"]["neg"])
    max_sent["neu"] = max(max_sent["neu"], data["senti"]["neu"])
    max_sent["amb"] = max(max_sent["amb"], data["senti"]["amb"])
```

```python
max_sent
```

```python
final_lst = []

for data in lst : 
    pos_score = float(data["senti"]["pos"]) / 28.0
    neg_score = float(data["senti"]["neg"]) / 28.0
    neu_score = float(data["senti"]["neu"]) / 28.0
    amb_score = float(data["senti"]["amb"]) / 28.0

    sum_of_scores = pos_score + neg_score + neu_score + amb_score

    if sum_of_scores == 0 : continue

    final_lst.append({
        "text" : data["text"],
        "senti" : {
            "pos" : pos_score / sum_of_scores,
            "neg" : neg_score / sum_of_scores,
            "neu" : neu_score / sum_of_scores,
            "amb" : amb_score / sum_of_scores
        }
    })
```
