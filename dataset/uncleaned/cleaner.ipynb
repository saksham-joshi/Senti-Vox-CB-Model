```python
from pandas import read_csv, read_excel
from pandas import isna, DataFrame
from typing import Dict, List
```

```python
ILLEGAL_CHARS_FOR_EXCEL_FILE = [
        '\x00',
        '\x01',
        '\x02',
        '\x03',
        '\x04',
        '\x05',
        '\x06',
        '\x07',
        '\x08',
        '\x0b',
        '\x0c',
        '\x0e',
        '\x0f',
        '\x10',
        '\x11',
        '\x12',
        '\x13',
        '\x14',
        '\x15',
        '\x16',
        '\x17',
        '\x18',
        '\x19',
        '\x1a',
        '\x1b',
        '\x1c',
        '\x1d',
        '\x1e',
        '\x1f'
]

def removeIllegalChars(__st : str) -> str :
    for illegal_char in ILLEGAL_CHARS_FOR_EXCEL_FILE :
        __st = __st.replace(illegal_char, "")
    return __st
```

```python
df = read_csv("Airline-Sentiment-2-w-AA.csv", encoding="latin-1")
```

```python
lst_of_cleaned_data : List[Dict[str , str | int]] = []
```

```python
def getSentimentNumber(__sentiment : str | None) -> int :

    if __sentiment is None : raise Exception("Don't you dare to paas a null/none value to me")
    
    __sentiment = __sentiment.lower().strip()

    if __sentiment == "positive" : return 1

    elif __sentiment == "neutral" : return 0

    elif __sentiment == "negative" : return 2

    raise Exception("Found invalid sentiment value '" + __sentiment + "'")
```

```python


for i in range(df.__len__()) :
    cur_row = df.iloc[i]

    comment_text = cur_row["text"].strip()

    if(comment_text.__len__()) == 0 :continue

    lst_of_cleaned_data.append({
        "text" : comment_text,
        "sentiment" : getSentimentNumber(cur_row["airline_sentiment"])
    })
```

```python
DataFrame(lst_of_cleaned_data).to_excel("cb-data-v1.xlsx", index=False)
```

```python
# 1 Negative
# 3 Neutral
# 5 Positive

df = read_csv("Apple-Twitter-Sentiment-DFE.csv", encoding="latin-1")
```

```python
df.head()
```

```python
lst_of_cleaned_data.clear()

sentiment_number_dt = {
    1 : "negative",
    3 : "neutral",
    5 : "positive"
}

for i in range(df.__len__()) :
    cur_row = df.iloc[i]

    comment_text = cur_row["text"].strip()

    if comment_text.__len__() == 0 or cur_row["sentiment"] == "not_relevant" :continue

    lst_of_cleaned_data.append({
        "text" : comment_text,
        "sentiment" : getSentimentNumber(sentiment_number_dt.get(int(cur_row["sentiment"])))
    })
```

```python
len(lst_of_cleaned_data)
```

```python
DataFrame(lst_of_cleaned_data).to_excel("cb-data-v2.xlsx", index=False)
```

```python
df = read_csv("IMDB-Dataset.csv", encoding="utf-8")
```

```python
df.head()
```

```python
lst_of_cleaned_data.clear()

# Source - https://stackoverflow.com/a/42365230
# Posted by user4896331
# Retrieved 2026-03-14, License - CC BY-SA 3.0

# df = df.apply(lambda x: x.encode('unicode_escape').decode('utf-8') if isinstance(x, str) else x)

for i in range(df.__len__()) :
    cur_row = df.iloc[i]

    comment_text = str(cur_row["review"]).replace("<br/>", "\n").replace("<br />", "\n").replace("<br>", "\n").replace("\t", " ").strip()
    comment_text = removeIllegalChars(comment_text).strip()
    if comment_text.__len__() == 0 :continue

    lst_of_cleaned_data.append({
        "text" : comment_text,
        "sentiment" : getSentimentNumber(cur_row["sentiment"])
    })
```

```python
DataFrame(lst_of_cleaned_data).to_excel("cb-data-v3.xlsx", index=False)
```
