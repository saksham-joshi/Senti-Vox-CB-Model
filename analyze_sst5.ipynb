```python
from json import loads, dump
from typing import Dict, Set, List
```

```python
lst = []
with open("dataset/uncleaned/sst5.jsonl", "r", encoding="utf-8") as file :
    lst.extend([loads(line) for line in file ])
```

```python
label_to_labet_text : Dict[int, Set[str]] = {}

for data in lst :

    label_set = label_to_labet_text.get(int(data['label']))

    if label_set is None : label_to_labet_text[int(data['label'])] = set([str(data['label_text'])])

    else : label_set.add(str(data['label_text']))
```

```python
label_to_labet_text
```

```python
normalized_labels = {
    0 : {
        "pos" : 0.0,
        "neg" : 1.0,
        "neu" : 0.0,
        "amb" : 0.0
    },
    1 : {
        "pos" : 0.2,
        "neg" : 0.7,
        "neu" : 0.1,
        "amb" : 0.0
    },
    2 : {
        "pos" : 0.1,
        "neg" : 0.1,
        "neu" : 0.8,
        "amb" : 0.0
    },
    3 : {
        "pos" : 0.7,
        "neg" : 0.2,
        "neu" : 0.1,
        "amb" : 0.0
    },
    4 : {
        "pos" : 1.0,
        "neg" : 0.0,
        "neu" : 0.0,
        "amb" : 0.0
    }
}
```

```python
final_normalized_dataset : List[Dict[str, str | Dict[str, float]]] = []

from scripts.tokenizer import clean_text

for data in lst:
    normalized_text = clean_text(data['text']).strip()

    if normalized_text.__len__() == 0 : continue

    final_normalized_dataset.append({
        "text" : normalized_text,
        "senti" : normalized_labels[data['label']]
    })
```

```python
with open("dataset/cleaned/sst5-normalized.json", "w", encoding="utf-8") as file :
    dump(final_normalized_dataset, file, ensure_ascii=False, allow_nan= False, indent=None, separators=(",", ":"))
```
